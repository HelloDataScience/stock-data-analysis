```python
# HTTP 요청을 보내기 위해 requests 라이브러리를 가져옵니다.
import requests
```

```python
# 삼성전자 주식 정보를 가져올 네이버 금융 URL을 정의합니다.
url = 'https://finance.naver.com/item/main.naver?code=005930'
```

```python
# 정의된 URL로 GET 요청을 보내고 응답을 'res' 변수에 저장합니다.
res = requests.get(url)
```

```python
# HTTP 요청의 상태 코드를 확인하여 성공 여부를 확인합니다.
res.status_code
```

```python
# 응답에서 HTML 텍스트 내용을 가져옵니다.
res.text
```

```python
# 응답 HTML 텍스트의 처음 100자를 출력하여 미리 봅니다.
print(res.text[:100])
```

```python
# HTML 파싱을 위해 BeautifulSoup 라이브러리를 'BTS'라는 별칭으로 가져옵니다.
from bs4 import BeautifulSoup as BTS
```

```python
# 응답 텍스트와 'html.parser'를 사용하여 BeautifulSoup 객체를 생성합니다.
soup = BTS(res.text, 'html.parser')
```

```python
# CSS 선택자를 사용하여 주식 가격이 포함된 HTML 요소를 선택하고 결과를 표시합니다.
items = soup.select('#rate_info_krx > div > p.no_today > em > span')
items
```

```python
# 선택된 항목들 중 첫 번째 항목에서 텍스트 콘텐츠(주식 가격)를 추출합니다.
items[0].text
```
